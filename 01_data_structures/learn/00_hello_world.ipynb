# Run your first Python notebook cell

In this notebook, you will learn how to select the Python kernel, run a code cell, and find its output.

## What is a Jupyter notebook?

A **Jupyter notebook** is a document that combines explanations with code you can run. A **text cell**, like this one, contains instructions. A **code cell** contains instructions for the computer.

In this class, you will use notebooks in **Grader Than Workspace** to read explanations, watch demonstrations, run examples, and practice writing code. You can work through a lesson and try its examples in the same document.

A **kernel** is the program that runs your code. Select the kernel for the language used in this notebook: **Python**. The **output** is the result shown beneath a cell after you run it. Simply reading or editing a cell does not run it.

## Try it yourself

The code below displays a short greeting. You do not need to understand its punctuation yet; focus on selecting the kernel and running the cell.

1. Find **Select Kernel** in the notebook's upper-right corner. If a kernel name is already shown there, click that name instead.
2. Choose **Python Environments...**, then select the recommended **.venv** Python environment. In the demonstration, its label is **.venv (Python 3.10.12)**; your version number may differ. If the picker first shows recent kernels, choose the matching **.venv** entry or use **Select Another Kernel...** to reach **Python Environments...**.
3. Check that the upper-right selector now shows **.venv** with a Python version. This tells you which Python environment will run the cell.
4. Move your pointer over the code cell below. Click the small triangular **Run** button at the cell's left edge. Give Python a moment to start; you do not need **Run All** at the top of the notebook.
5. Look directly below the cell for its output.

````python
print("Hello, world!")
````

## Check the result

After the cell runs, you should see this output directly beneath it:

```text
Hello, world!
```

If you see that greeting without an error, you have successfully run the cell. You can run this greeting cell again to display the same message.

## If the cell does not run

Give the kernel a moment to start. If the notebook asks you to select a kernel, follow the steps above and choose the Python kernel. If that kernel is missing or an error remains, ask your instructor for help. You do not need to install anything to complete this activity.
