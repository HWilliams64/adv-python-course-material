# Function and Method Overloading in Python

Python does not support traditional function or method overloading in the same way as languages such as Java or C++. Instead, Python usually uses default arguments, `*args`, `**kwargs`, and explicit type checks when one function needs to handle different input patterns.

## Overview

In traditional overloading, a language may allow several functions with the same name but different parameter lists. Python keeps the most recent definition of a function name, so defining the same function twice does not create two overloads.

````python
def greet(name):
    print(f"Hello, {name}")

def greet(first_name, last_name):
    print(f"Hello, {first_name} {last_name}")

# The second definition replaced the first one.
greet("Ada", "Lovelace")

# Uncomment the next line to see the error:
# greet("Ada")
````

Instead of writing multiple functions with the same name, Python programmers usually design one flexible function that can handle optional or variable inputs.

## Using Default Arguments for Overloading

Default arguments let a function be called with fewer arguments than it is defined to accept. This is one of the most common ways to simulate overloading in Python.

### Example with Default Arguments

````python
def greet(name, msg="Hello"):
    print(f"{msg}, {name}")

greet("Alice")
greet("Bob", "Goodbye")
````

- **`greet` function**: Accepts a required name and an optional message.
- **Default value**: If no message is provided, `msg` defaults to `"Hello"`.
- This simulates overloading by allowing the function to be called with different numbers of arguments.

`*args` allows a function to accept any number of positional arguments. `**kwargs` allows a function to accept any number of keyword arguments. Together, they make a function flexible enough to handle many input patterns.

### Example with `*args` and `**kwargs`

````python
def profile(name, *activities, **details):
    print(f"Name: {name}")
    print("Activities:", ", ".join(activities))
    for key, value in details.items():
        print(f"{key}: {value}")

profile("Alice", "Coding", "Reading", age=30, language="Python")
````

- **`name`** is a normal required parameter.
- **`*activities`** collects extra positional arguments into a tuple.
- **`**details`** collects extra keyword arguments into a dictionary.
- This creates one flexible function instead of several overloaded versions.

## Method Overloading with Classes

In classes, method overloading can also be simulated with default arguments or `*args` and `**kwargs`. This lets one method respond to different numbers of inputs.

### Example in Class Context

````python
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30, 40))
````

- **`add` method**: Accepts any number of arguments and returns their sum.
- A single method can handle several calling patterns, which gives behavior similar to overloading.

## When to Use Each Strategy

- Use **default arguments** when only a few parameters are optional.
- Use **`*args`** when the number of positional inputs can vary.
- Use **`**kwargs`** when callers may provide named options.
- Use clear parameter names and validation so flexible functions remain readable.

````python
def calculate_total(*prices, discount=0):
    subtotal = sum(prices)
    return subtotal - (subtotal * discount)

print(calculate_total(10, 20, 30))
print(calculate_total(10, 20, 30, discount=0.10))
````

## Conclusion

Python does not support traditional function or method overloading, but default arguments, `*args`, and `**kwargs` provide a flexible alternative. These tools let you write functions and methods that handle a range of input scenarios while keeping one clear function name.

def average(*numbers, round_to=None):
    if not numbers:
        return 0

    result = sum(numbers) / len(numbers)

    if round_to is not None:
        return round(result, round_to)

    return result

class Course:
    def __init__(self, title):
        self.title = title
        self.students = []

    def add_students(self, *names):
        for name in names:
            self.students.append(name)

    def describe(self):
        print(f"Course: {self.title}")
        print("Students:")
        for student in self.students:
            print(f"- {student}")

course = Course("Advanced Python")
course.add_students("Ada", "Grace")
course.add_students("Katherine")
