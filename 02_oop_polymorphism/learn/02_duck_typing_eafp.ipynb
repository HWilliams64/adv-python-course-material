Duck typing is a concept in Python that focuses on what an object can do, rather than what type or class the object belongs to. If an object has the methods or attributes your code needs, your code can use it.

### Example of Duck Typing

In this example, the function does not check whether the object is a `Duck`. It only assumes the object has a `quack()` method.

````python
class Duck:
    def quack(self):
        print("Quack, quack!")

class Person:
    def quack(self):
        print("I'm quacking like a duck!")

def make_it_quack(duck):
    duck.quack()

duck = Duck()
person = Person()

make_it_quack(duck)
make_it_quack(person)
````

- **The `make_it_quack` function** expects an object that has a `quack()` method.
- **The `Duck` and `Person` classes** are different types, but both work because they provide the behavior the function needs.
- This is polymorphism in practice: different objects can be used through a shared behavior.

EAFP is a Python programming principle that means you try to do the operation first and handle the exception if it fails. This is different from Look Before You Leap (LBYL), where you check every condition before attempting the operation.

### Example of EAFP

The function below tries to open a file. If the file does not exist, the `except` block handles the problem.

````python
def read_file(file_name):
    try:
        with open(file_name) as file:
            print(file.read())
    except FileNotFoundError:
        print("The file does not exist")

read_file("example.txt")
````

- **Try-except block**: The code attempts the operation and handles the specific error that may occur.
- **Cleaner control flow**: EAFP can make code easier to read because the normal action is written first.
- **Specific exceptions**: Catch the exception you expect instead of catching every possible error.

### EAFP Compared with LBYL

Both versions below try to get the first item from a list. The LBYL version checks first. The EAFP version tries first and handles the problem if the list is empty.

````python
def first_item_lbyl(items):
    if len(items) > 0:
        return items[0]
    return None

def first_item_eafp(items):
    try:
        return items[0]
    except IndexError:
        return None

print(first_item_lbyl([10, 20, 30]))
print(first_item_lbyl([]))

print(first_item_eafp([10, 20, 30]))
print(first_item_eafp([]))
````

## How EAFP Relates to Polymorphism

EAFP and duck typing both support Python's flexible approach to polymorphism. Duck typing lets objects work together based on shared behavior, and EAFP lets your code try that behavior directly while handling the case where the behavior is missing or fails.

````python
class CSVReport:
    def export(self):
        return "Exporting report as CSV"

class PDFReport:
    def export(self):
        return "Exporting report as PDF"

class DraftNote:
    pass

def export_document(document):
    try:
        print(document.export())
    except AttributeError:
        print("This object cannot be exported")

export_document(CSVReport())
export_document(PDFReport())
export_document(DraftNote())
````

- **Flexibility and reusability**: The function works with any object that provides an `export()` method.
- **Polymorphic behavior**: Different report objects respond to the same method call in their own way.
- **Graceful failure**: EAFP allows the function to handle objects that do not support the expected behavior.
